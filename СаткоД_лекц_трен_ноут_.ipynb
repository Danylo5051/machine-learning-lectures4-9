13. Лінійна алгебра
Матричний добуток

```python
A = np.array([
    [1, 2, 3],
    [0, 1, 2]
])

B = np.array([
    [ 1, 2],
    [ 0, 4],
    [-1, 0]
])

C = A @ B
print(C)
```

Діагональ, транспонування та визначник

```python
A = np.array([
    [ 1.,  0., 3.],
    [-1., -1., 2.],
    [ 4.,  7., 2.]
])

print("diag:", np.diag(A))
print("\nA.T =\n", A.T)

det = np.linalg.det(A)
print("\ndet(A) =", det)
print("det близький до -25:", np.isclose(det, -25.0))
```

Розв’язання системи Ax = b

```python
A = np.array([
    [1.,  2.,  5.],
    [1., -1.,  3.],
    [3., -6., -1.]
])

b = np.array([-9., 2., 25.])

x = np.linalg.solve(A, b)

print("x =", x)
print("Перевірка A @ x ≈ b:", np.allclose(A @ x, b))
```

Для розв’язання системи не потрібно явно обчислювати A⁻¹.
У числових обчисленнях перевага надається np.linalg.solve(A, b).

Метод найменших квадратів

```python
X = np.array([
    [1., 1.],
    [1., 2.],
    [1., 3.],
    [1., 4.]
])

y = np.array([2.1, 3.9, 6.2, 8.1])

beta, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)

print("beta:", beta)
print("residuals:", residuals)
print("rank:", rank)
print("singular values:", s)
```

Власні значення та власні вектори

```python
A = np.array([
    [2., 1.],
    [1., 2.]
])

values, vectors = np.linalg.eigh(A)

print("Власні значення:", values)
print("Власні вектори:\n", vectors)
```

SVD

```python
A = np.array([
    [3., 1.],
    [1., 3.],
    [1., 1.]
])

U, s, Vt = np.linalg.svd(A, full_matrices=False)

print("U =\n", U)
print("\ns =", s)
print("\nVt =\n", Vt)
```

18. Узагальнюючий приклад: підготовка даних і лінійний прогноз

```python
# 1. Вхідні дані: два фактори та цільова змінна
X = np.array([
    [10., 2.],
    [12., 3.],
    [15., 4.],
    [18., 5.],
    [20., 6.]
])

y = np.array([21., 25., 31., 36., 40.])

# 2. Фільтрація
mask = np.isfinite(X).all(axis=1) & np.isfinite(y)
X = X[mask]
y = y[mask]

# 3. Стандартизація ознак
mu = X.mean(axis=0)
sigma = X.std(axis=0)
X_std = (X - mu) / sigma

# 4. Додаємо стовпець одиниць для вільного члена
X_design = np.column_stack([
    np.ones(X_std.shape[0]),
    X_std
])

# 5. Оцінювання параметрів методом найменших квадратів
beta, *_ = np.linalg.lstsq(X_design, y, rcond=None)

# 6. Прогноз і MSE
y_pred = X_design @ beta
mse = np.mean((y - y_pred) ** 2)

print("beta =", beta)
print("y_pred =", y_pred)
print("MSE =", mse)
```

19. Підсумкові тренувальні завдання
Створіть матрицю X форми (5, 4) із випадкових чисел.
Обчисліть середнє та стандартне відхилення кожної ознаки.
Стандартизуйте матрицю через broadcasting.
Виберіть рядки, у яких значення першої ознаки додатне.
Додайте до матриці новий стовпець одиниць.
Створіть вектор ваг і обчисліть прогноз X @ w.
Для заданого y_true обчисліть MSE.
Перевірте, чи немає в масивах NaN або inf.

```python
import numpy as np

rng = np.random.default_rng(42)

X = rng.normal(loc=5.0, scale=2.0, size=(5, 4))

mu = X.mean(axis=0)
sigma = X.std(axis=0)

X_std = (X - mu) / sigma

mask = X_std[:, 0] > 0
X_filtered = X_std[mask]

X_design = np.column_stack([
    np.ones(X_filtered.shape[0]),
    X_filtered
])

w = np.array([1.0, 0.5, -0.5, 0.2, 0.1])
y_pred = X_design @ w

y_true = rng.uniform(0, 2, size=y_pred.shape[0])
mse = np.mean((y_true - y_pred) ** 2)

is_valid = np.isfinite(X_design).all() and np.isfinite(y_true).all()

print("Відфільтрована матриця з одиницями (X_design):\n", X_design)
print("\nПрогноз (y_pred):", y_pred)
print("MSE =", mse)
print("Усі значення скінченні (без NaN/inf):", is_valid)
```
